```python
plt.figure(figsize=(10, 5))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False, cmap="viridis")
plt.title("Missing Values in the Titanic Data (yellow = missing)")
plt.xlabel("Column")
plt.ylabel("Passengers")
plt.show()
```

Question 4

```python
plt.figure(figsize=(8, 5))
plt.hist(df["flipper_length_mm"], bins=20)
plt.title("Lenght of the flipper ")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Number of penguins")
plt.show()
```

Question 5
Bar Chart

```python
penguins_count = df["island"].value_counts()
print(penguins_count)
```

```python
plt.figure(figsize=(8, 5))
bars = plt.bar(penguins_count.index, penguins_count.values)
plt.bar_label(bars)
plt.title("Number of Penguins in Each Island")
plt.xlabel("Island")
plt.ylabel("Number of Penguins")
plt.show()
```

Question 6

Scatter plot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="bill_length_mm", y="bill_depth_mm", hue="species")
plt.title("Bill length vs Bill Depth, Coloured by Species")
plt.xlabel("Bill Length (mm)")
plt.ylabel("Bill Depth (mm)")
plt.show()
```
